```python
import os

file_organic = '/home/anya/Documents/diploma_datasets/yambda_organic/yambda_organic.inter'
file_random  = '/home/anya/Documents/diploma_datasets/yambda_random/yambda_random /yambda_random.inter'
file_rec     = '/home/anya/Documents/diploma_datasets/yambda_rec/yambda_algo.inter'

print(f"Organic найден: {os.path.exists(file_organic)}")
print(f"Random найден:  {os.path.exists(file_random)}")
print(f"Rec найден:     {os.path.exists(file_rec)}")
```

```python
import pandas as pd

datasets = {
    'Yambda-Organic': file_organic,
    'Yambda-Random': file_random,
    'Yambda-Rec': file_rec
}

user_col = 'user_id:token'
item_col = 'item_id:token'


results = []

```

```python

for name, path in datasets.items():
    

    df = pd.read_csv(path, sep='\t', usecols=[user_col, item_col])
    
   
    num_users = df[user_col].nunique()
    num_items = df[item_col].nunique()
    num_interactions = len(df)
    
    
    max_possible = num_users * num_items
    density = (num_interactions / max_possible) * 100 if max_possible > 0 else 0
    
    
    results.append({
        'Версия датасета': name,
        'Пользователи': f"{num_users:,}".replace(',', ' '),
        'Треки': f"{num_items:,}".replace(',', ' '),
        'Взаимодействия': f"{num_interactions:,}".replace(',', ' '),
        'Плотность матрицы': f"{density:.2f}%"
    })
    

```

```python

final_table = pd.DataFrame(results)
display(final_table)
```

```python
output_csv = '/home/anya/Documents/diploma_datasets/dataset_characteristics.csv'

final_table.to_csv(output_csv, index=False, encoding='utf-8')
```
